```python
import pandas as pd

df = pd.read_csv("Sample - Superstore.csv.csv", encoding='latin1')

print(df.head())
```

```python
print(df.shape)
print(df.columns)
print(df.info())
```

```python
# Total Sales
print("Total Sales:", df['Sales'].sum())

# Total Profit
print("Total Profit:", df['Profit'].sum())

# Sales by Category
print("\nSales by Category:")
print(df.groupby('Category')['Sales'].sum())

# Profit by Region
print("\nProfit by Region:")
print(df.groupby('Region')['Profit'].sum())
```

```python
import matplotlib.pyplot as plt

# Sales by Category
category_sales = df.groupby('Category')['Sales'].sum()

category_sales.plot(kind='bar')

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.show()
```

```python
# Profit by Region Chart

region_profit = df.groupby('Region')['Profit'].sum()

region_profit.plot(kind='bar')

plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Profit")

plt.show()
```

```python
# Convert Order Date to datetime
df['Order Date'] = pd.to_datetime(df['Order Date'])

# Monthly Sales Trend
monthly_sales = df.groupby(df['Order Date'].dt.to_period('M'))['Sales'].sum()

# Plot chart
monthly_sales.plot(kind='line', figsize=(12,5))

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
# BUSINESS INSIGHTS

# 1. Technology category generated the highest sales revenue.

# 2. West region achieved the highest profit among all regions.

# 3. Central region showed comparatively lower profitability and may require business improvement strategies.

# 4. Monthly sales trends indicate fluctuations in customer demand throughout the year.

# 5. Furniture category generated high sales but comparatively lower profit margins.

# RECOMMENDATIONS

# 1. Increase investment in Technology products.

# 2. Improve marketing and operations in Central region.

# 3. Analyze low-profit products and optimize discounts.

# 4. Focus on seasonal sales trends for better inventory planning.
```
